## Baseline for Random Submission - benchmark

```python
import random
import librosa


random.seed(23)


class Submission(object):
    def __init__(self):
        self.root_path = '/kaggle/input/competitions/asr-2026-spoken-numbers-recognition-challenge'
        self.test_df = pd.read_csv(f'{self.root_path}/test.csv')
        self.save_path = '/kaggle/working'

    def score_file(self, filepath: str = None):
        audio_samples, _ = librosa.load(filepath, sr=16_000)  # make sure you trained model for 16 kHz input, otherwise do resampling here

        # <PLACEHOLDER>
        # Do your NN inference and decoding for audio_samples here

        # I use random number for a submission
        transcription = random.randint(1_000, 999_999)
        assert type(transcription) == int
        return transcription
    
    def prepare_submission(self):
        self.test_df['transcription'] = self.test_df['filename'].apply(
            lambda x: self.score_file(f'{self.root_path}/{x}')
        )

    def save_submission(self, path: str =None):
        if not path:
            path = f'{self.save_path}/submission.csv'
        self.test_df[['filename', 'transcription']].to_csv(path, index=False)
```

```python
# create submission
submission = Submission()

# score each file with random value
submission.prepare_submission()

submission.test_df.head()
```

```python
# save results to csv for submission
submission.save_submission()
```
